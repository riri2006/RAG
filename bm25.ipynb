```python
from rank_bm25 import BM25Okapi
```

## Step 1: create documents

```python
Documents=[
    "Apple is red in colour.",
    "Banana is yellow in colour.",
    "Strawberry is red in colour.",
    "Mango and banana both are yellow in colour."
]
```

## Step 2: Tokenise documents:
#### Documents are needed to tokenise so that computer can individually process them

```python
tokens=[]
for doc in Documents:
    t=doc.lower().split()
    tokens.append(t)
bm25= BM25Okapi(tokens)
```

## Step 3: Query and tokenisation
#### Write query and again tokenise it 

```python
query="apple mango both colour "
token_query=query.lower().split()
```

## Step 4: print bm25 scores
#### It will give an array of all documents with some numeric values that indicated which document is highly relevant

```python
scores= bm25.get_scores(token_query)
print(scores)
```
